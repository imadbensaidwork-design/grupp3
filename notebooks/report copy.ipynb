```python
# här tar vi fram en funktion för att läsa in csv-filer
# koden nedan inspirerad från uppgift 17 i övningsuppgifter för vecka 3

import csv

category_revenue = {} 

# dessa ska göras till funktioner och sparas i ioutils.py

# den här räknar ut  vilken kategori som säljer bäst

with open("../data/ecommerce_sales.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    for row in reader:
        category = row["category"] # Hämtar produtkategorin (category)
        revenue = float(row["revenue"]) # Hämtar intäkt som float (revenue)

        if category in category_revenue: # om kategorin redan finns i vår dict category_revenue
            category_revenue[category] += revenue # lägg då till beloppet från revenue
        else:                                    # annars
            category_revenue[category] = revenue # lägg till ny kategori (med tillgörande revenue)

# den här räknar ut vilken stad som säljer bäst 
city_revenue = {}

with open("../data/ecommerce_sales.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    for row in reader:
        city = row["city"] # Hämtar produtkategorin (category)
        revenue = float(row["revenue"]) # Hämtar intäkt som float (revenue)

        if city in city_revenue: # om kategorin redan finns i vår dict category_revenue
            city_revenue[city] += revenue # lägg då till beloppet från revenue
        else:                                    # annars
            city_revenue[city] = revenue # lägg till ny kategori (med tillgörande revenue)

# print (category_revenue)
# print (city_revenue)
# max(category_revenue.values())

# Source - https://stackoverflow.com/a/5326622
# Posted by Hugh Bothwell
# Retrieved 2026-10-19, License - CC BY-SA 2.5

lst = [{'price': 99, 'barcode': '2342355'}, {'price': 88, 'barcode': '2345566'}]

maxPricedItem = max(lst, key=lambda x:x['price'])
minPricedItem = min(lst, key=lambda x:x['price'])


```
