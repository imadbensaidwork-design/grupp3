## Data Exploration
- Undersöka dataset struktur och datakvalitet innan dataresning 

```python
from pathlib import Path
import pandas as pd

print ("Working Directory: " + str(Path.cwd()))
```

```python
# hitta projekt root mappen
root = Path.cwd()
# Men detta gör vi om vi beffinner oss i notebooks map då går vi upp en nivå 
# Då kan köra koden från notebooks mappen och ändå hitta data mappen där dataset ligger
if root.name == "notebooks":
    root = root.parent 
```

```python
# Läs in dataset raw med Pandas
df_raw = pd.read_csv(root / "data" / "ecommerce_sales.csv")
df_raw.head()
```

#### Dataset struktut 
- Undersöker data struktur för att få en överblick av data sets 

```python
# Datasets storlek 
print("DATASET SHAPE")
print(df_raw.shape[0], "rows and", df_raw.shape[1], "columns")
```

```python
# Kolla datatyper
print("DATASET DATATYPES")
print(df_raw.dtypes)
```

```python
#Kolla dataset info
print("DATASET INFO")
print(df_raw.info())
```

```python
# Unika värden i varje kolumn
print("UNIKA VÄRDEN")
print(df_raw.nunique())
```

```python
# Beskrivande statistik för numeriska kolumner
print("DESCRIPTIVE STATISTICS")
print(df_raw.describe())
```

#### Observationen 
- Dataset innehåller **2.500 rader och 7 kolumner**
- Dataset innehåller inga **null** value enligt df_raw.info()
- Kolumner **order_id** har 2.500 unika värden, det man ser är att varje rad är en separat orden 
- Kolumner **city har 5** unika värden och **category har 6**
- Kolumner **date** är lagrad som str och behöver kolla om det är värt att konvertera den till datetime
  

### Datakvalitet
Undersöker för att identifiera problement innan dataresning

```python
# Saknade värden i datasetet
print("MISSING VALUES")
print(df_raw.isnull().sum())
```

```python
# Duplikatvärden i datasetet
print("DUPLICATE VALUES")
print("Nummer av duplikatvärden: ", df_raw.duplicated().sum())
print("Duplikatvärden i order_id kolumnen: ", df_raw['order_id'].duplicated().sum())
```

```python
# Kategoriska kolumner
print("UNIQUE CATEGORICAL VALUES")
print(df_raw["category"].unique())
```

```python
print("UNIQUE CITY VALUES")
print(df_raw["city"].unique())
```

- Dataset innehåller inga saknade value, inga duplikering och inga "typo" i city eller category kolummer 

```python
# Datumvalidering
print("DATE VALIDATION")
# Konvertera date kolumnen från str till datetime format och hantera felaktiga datum
dates = pd.to_datetime(df_raw['date'], errors='coerce')

print("Antal ogiltiga datum: ", dates.isnull().sum())
print("Antal unika datum: ", dates.nunique())
print("Tidigaste datum: ", dates.min())
print("Senaste datum: ", dates.max())
```
